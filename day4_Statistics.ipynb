```python
import numpy as np
data =np.array([2,4,4,4,5,5,7,9],dtype=float)
# mean- average value
mean=data.mean()
print("Mean:", mean)
# variance- average squared distance from mean
variance=data.var()
print("Variance:",variance)
#Standard deviation- square root of variance
std=data.std()
print("Std:",std)
#verify manually
manual_var=((data-mean)**2).mean()
print("Manual variance:", manual_var)
print("Same?",np.allclose(variance,manual_var))
```

```python
import numpy as np
data=np.array([2,4,4,4,5,7,9],dtype=float)
mean=data.mean()
std=data.std()
print("Mean:",mean)
print("Std:",std)
print()
# How far each value from the mean
distances=data-mean
print("Distances form the mean:",distances)
# Values within 1 std of mean
within_1std=data[np.abs(data-mean)<=std]
print("Within 1 std:",within_1std)
# Normalise - Zero mean , unit variance
normalised = (data-mean)/std
print("\n Normalised:",normalised)
print("New mean:",normalised.mean().round(10))
print("New std:",normalised.std())
```

```python
import numpy as np
embeddings=np.random.randn(5,8)
print(embeddings)
# Without keepdims
mean = embeddings.mean(axis=1)
print("mean:",mean)
print("Without keepdims",mean.shape)
# With keepdims
mean=embeddings.mean(axis=1,keepdims=True)
print("With keepdims:",mean.shape)
# keepdims=True = [:, np.newaxis]
#Both do the same thing - keep the dimension alive as size 1, so broadcasting can work.
```

```python
import numpy as np

# Create random embeddings
# 5 tokens, each represented by 8 numbers
# randn = random numbers from normal distribution
# *10 = scale up so numbers are bigger and easier to see
np.random.seed(42)  # seed makes random numbers same every time you run
embeddings = np.random.randn(5, 8) * 10

print("Shape:", embeddings.shape)  # should be (5, 8)

# ── Before normalisation ──────────────────────────────
# axis=1 means compute mean ACROSS each row (each token)
# keepdims=True keeps shape (5,1) instead of (5,)
mean = embeddings.mean(axis=1, keepdims=True)   # shape (5, 1)
std  = embeddings.std(axis=1,  keepdims=True)   # shape (5, 1)

print("\nBefore normalisation:")
print("Mean per token:", embeddings.mean(axis=1).round(2))  # one mean per token
print("Std per token: ", embeddings.std(axis=1).round(2))   # one std per token

# ── Normalise ─────────────────────────────────────────
# Formula: (value - mean) / std
# This makes every token have mean=0 and std=1
# Same formula used inside every transformer block (Layer Norm)
normalised = (embeddings - mean) / std

print("\nAfter normalisation:")
# round(10) removes floating point noise like 0.000000000001
print("Mean per token:", normalised.mean(axis=1).round(10))  # should be 0.0
print("Std per token: ", normalised.std(axis=1).round(2))    # should be 1.0
```
