```python
# Matrix Multiplication
import numpy as np
A=np.array([[1,2],
           [3,4],
           [5,6]])
B=np.array([[7,8,9],
           [10,11,12]])
result1=np.dot(A,B)
result2=A@B
result3=np.matmul(A,B)
print("Shape of A:",A.shape)
print("Shape of B: ",B.shape)
print("Shape of Result2: ",result2.shape)
print("\nResult:\n",result2)
print("\n all same ?",np.allclose(result1,result2))
```

```python
#In transformers, matrix multiplication is how attention scores are computed:
import numpy as np

# Simulate a tiny attention operation
# 4 tokens, each with embedding size 8
seq_len = 4
embed_dim = 8

# Random token embeddings
X = np.random.randn(seq_len, embed_dim)   # (4, 8)

# Weight matrices for Q, K, V
Wq = np.random.randn(embed_dim, embed_dim)  # (8, 8)
Wk = np.random.randn(embed_dim, embed_dim)  # (8, 8)

# Compute queries and keys
Q = X @ Wq   # (4, 8)
K = X @ Wk   # (4, 8)

# Attention scores — every token attends to every other token
scores = Q @ K.T   # (4, 8) @ (8, 4) = (4, 4)

print("X shape:", X.shape)
print("Q shape:", Q.shape)
print("K shape:", K.shape)
print("Scores shape:", scores.shape)
print("\nAttention scores:\n", scores.round(2))
```
